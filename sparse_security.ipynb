```python
from estimator import *
import math
from sage.all import oo, log
import json
import numpy as np
from estimator.nd import SparseBinary, DiscreteGaussian
from monte_carlo_estimator import estimate_delta_montecarlo
from estimator.lwe_parameters import LWEParameters
```

```python
batch_size = 10**6
N = 2**12
h = 39
B = 2**7
q = 2**64
sigma = 2**49
```

```python
h = 39
N = 2**11
B = 2**7
sigma = 2**49
B1 = LWEParameters(
    n=2**11,
    q=q,
    Xs=SparseBinary(h),
    Xe=DiscreteGaussian(stddev=sigma),
    tag="B1",
)
r = LWE.estimate(B1)
security = min([
    math.log2(attack["rop"]) for attack in r.values()]
)
delta = estimate_delta_montecarlo(N, h, B, batch_size)
print(f"B1 security : {security-delta}")
```
